```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = 'https://raw.githubusercontent.com/lejejefr/ECM_2526_FinalProject/refs/heads/main/Data/RawData.csv'
df = pd.read_csv(url)
df.head()
```

```python
date_range = pd.to_datetime(df["Date"]).max() - pd.to_datetime(df["Date"]).min()
print(date_range)
```

```python
plt.figure(figsize=(8, 8))
sns.scatterplot(
    data=df[df['Longitude'] > -90],
    x="Longitude",
    y="Latitude",
    s=10
)

plt.title("Localisation des incidents (sans outlier)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

```python
plt.figure(figsize=(8, 8))
sns.kdeplot(
    data=df[df['Longitude'] > -90],
    x="Longitude",
    y="Latitude",
    fill=True,
    cmap="Reds"
)
plt.title("Densité spatiale des crimes (KDE)")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df['Date'] = pd.to_datetime(df['Date'])

df['Hour'] = df['Date'].dt.hour
df['Day_of_Week'] = df['Date'].dt.day_name()
df['Day_of_Year'] = df['Date'].dt.dayofyear

days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 18))

sns.histplot(data=df, x='Hour', bins=24, color='skyblue', ax=axes[0])
axes[0].set_title('Distribution des crimes par Heure de la journée')
axes[0].set_xticks(range(0, 24))

sns.countplot(data=df, x='Day_of_Week', order=days_order, palette='viridis', ax=axes[1])
axes[1].set_title('Nombre de crimes par Jour de la semaine')

sns.histplot(data=df, x='Day_of_Year', bins=52, color='salmon', ax=axes[2])
axes[2].set_title('Évolution des crimes sur l\'année (par jour)')

plt.tight_layout()
plt.show()
```
